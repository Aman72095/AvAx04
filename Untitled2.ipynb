```python
import pandas as pd

df = pd.read_csv("crops_phytochemicals.csv")
print(df.head())
```

```python
X = df.drop(["Crop", "AntiCancer"], axis=1)
y = df["AntiCancer"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

y_pred = model.predict(X_test)

print("Classification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt

feat_importances = pd.Series(model.feature_importances_, index=X.columns)
feat_importances.sort_values().plot(kind="barh", figsize=(8,5))
plt.title("Feature Importance in Anti-Cancer Prediction")
plt.show()
```

```python
new_crop = pd.DataFrame({
    "Polyphenols": [130],
    "Terpenoids": [70],
    "Alkaloids": [7],
    "Flavonoids": [90]
})

prediction = model.predict(new_crop)[0]
print("Predicted Anti-Cancer Potential:", "Yes" if prediction == 1 else "No")
```
